```python
"""
Script to plot mutational burden per sample
Input: mutation_burden.txt
Output: Barplot PNG
"""
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# Define input and output paths
input_path = "analysis_results/mutation_burden/mutation_burden.txt"
output_path = "analysis_results/mutation_burden/mutation_burden.png"

# Create output directory if needed
os.makedirs(os.path.dirname(output_path), exist_ok=True)

# Load mutation burden data (generated by bash script (compute_mutation_burden.sh)
mutation_burden = pd.read_csv(input_path, sep=r"\s+", names=["Sample", "Mutations"])

# Sort by mutation count (descending)
mutation_burden = mutation_burden.sort_values(by="Mutations", ascending=False)

# Set up the plot
plt.figure(figsize=(10, 5))
sns.barplot(y="Sample", x="Mutations", data=mutation_burden,
            palette="crest_r", hue="Sample", legend=False)

# Aesthetics
plt.title("Mutation Burden Per Sample")
plt.xlabel("Mutation Count")
plt.ylabel("Sample")

# Save the plot
plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()
```
