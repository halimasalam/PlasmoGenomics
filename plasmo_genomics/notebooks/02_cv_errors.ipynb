```python
# PCA ADMIXTURE Cross-Validation Error

### Find the best K for admixture by plotting cv_errors
import matplotlib.pyplot as plt

# Read CV errors
with open("analysis_results/admixture/files/cv_errors.txt", "r") as f:
    lines = f.readlines()

# Extract K values and errors
K_values = []
CV_errors = []
for line in lines:
    parts = line.strip().split(":")
    if len(parts) == 2:
        K = int(parts[0].split("=")[1].strip(")"))
        error = float(parts[1])
        K_values.append(K)
        CV_errors.append(error)

# Plot
plt.figure(figsize=(8,5))
plt.plot(K_values, CV_errors, marker='o', linestyle='-', color='b')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("CV Error")
plt.title("ADMIXTURE Cross-Validation Error")
plt.xticks(K_values)
plt.grid()
plt.savefig("analysis_results/admixture/plots/Admixture_cross-validation-error.png", dpi=300, bbox_inches='tight')
plt.show()
```
