```python
## Generate phylogeny tree
from ete3 import Tree, TreeStyle

# Load tree from Newick format (treefile)
t = Tree("analysis_result/phylogeny/files/high_quality_variants.phy.treefile", format=1)

# Style the tree
ts = TreeStyle()
ts.show_leaf_name = True  # Display leaf labels

# Render and save as PNG
t.render("analysis_result/phylogeny/plots/high_quality_variants_tree.png", w=800, tree_style=ts)
```
