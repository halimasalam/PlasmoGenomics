```python
# Plot: Total Mutations per Gene for Genes of Interest in the VCF
```

```python


import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Input/Output Paths 
input_file = "analysis_results/variants_analysis/files/genes_of_interest.tsv"
output_plot = "analysis_results/variants_analysis/plots/total_mutations_per_gene.png"

# Load and Preprocess Data
# Load mutation data with gene annotations
df_gene = pd.read_csv(input_file, sep="\t")

# Combine Gene Symbol and Annotation Type for better readability in the legend
df_gene["PmUG0_ID"] = df_gene.apply(
    lambda row: f"{row['PmUG0_ID']} ({row['Gene']}, {row['Annotation_type']})"
    if pd.notna(row["Gene"]) else f"{row['PmUG0_ID']} ({row['Annotation_type']})",
    axis=1
)

# Count Mutations per Gene per Sample
gene_counts = df_gene.groupby(["PmUG0_ID", "Sample"]).size().reset_index(name="MutationCount")

# Pivot the DataFrame to prepare for stacked bar plot
pivot_df = gene_counts.pivot(index="Sample", columns="PmUG0_ID", values="MutationCount").fillna(0)

# Plotting
fig, ax = plt.subplots(figsize=(50, 22))

# Create a stacked bar plot
pivot_df.plot(kind="bar", stacked=True, colormap="tab10", edgecolor="black", ax=ax)

# Add plot labels and formatting
ax.set_xlabel("Sample", fontsize=30, fontweight="bold")
ax.set_ylabel("Total Mutation Count", fontsize=30, fontweight="bold")
ax.set_title("Total Mutations Count per Isolate for Genes of Interest (Pm)", fontsize=32, fontweight="bold")

# Rotate x-tick labels for better readability
ax.tick_params(axis="x", rotation=45, labelsize=18)
ax.tick_params(axis="y", labelsize=18)

# Customize legend
ax.legend(
    title="Sample (Gene, Variant Classification)",
    bbox_to_anchor=(1, 1),
    prop={'size': 21},
    title_fontsize=20
)

# Save and show the plot
plt.tight_layout()
plt.savefig(output_plot, dpi=300, bbox_inches='tight')
plt.show()

print(f"✅ Mutation plot saved to: {output_plot}")
```
